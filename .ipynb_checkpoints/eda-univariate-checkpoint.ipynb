```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

url = "D:/04_IronHack_DA/IRONHACK_Labs/amz_uk_price_prediction_dataset.csv"
df = pd.read_csv(url)
```

```python
# Part 1: Understanding Product Categories
# Business Question: What are the most popular product categories on Amazon UK, and how do they compare in terms of listing frequency?

# Frequency Tables:
frequency_table = df['category'].value_counts()
frequency_table_top=frequency_table.head()

# Visualizations:

frequency_table_top.plot.barh()
frequency_table_top.plot.pie()

# The category "sports and outdoors" dominates all the other categories by a huge margin
```

```python
#Part 2: Delving into Product Pricing
#Business Question: How are products priced on Amazon UK, and are there specific price points or ranges that are more common?

# 1.Measures of Centrality:

mean_price = df['price'].mean().round(2)
median_price = df['price'].median().round(2)
mode_price = df['price'].mode().round(2)

# Average price point is is alomst 10 times of most common price

# 2.Measures of Dispersion:
variance_price = df['price'].var().round(2)
std_dev_price = df['price'].std().round(2)
min_price = df['price'].min().round(2)
max_price = df['price'].max().round(2)
range_price = (max_price - min_price).round(2)
quantiles_price = df['price'].quantile([0.25, 0.5, 0.75])
display(variance_price, std_dev_price, min_price, max_price, range_price, quantiles_price)

# 50% of the products are priced between 9.99 and 45.99, however the max price is 100000.0 and thats a very huge outlier, which means the graph is skewed right


# 3.Visualizations:

skewness_price = df['price'].skew()
kurtosis_price = df['price'].kurtosis()

skewness_price, kurtosis_price
sns.histplot(df['price'], kde=True, bins=50, color="red")

# It is hard to read the graph because of huge difference in price frequencies

sns.boxplot(data = df['price'], color="lightblue").
```

```python
#Part 3: Unpacking Product Ratings
#Business Question: How do customers rate products on Amazon UK, and are there any patterns or tendencies in the ratings?

#Measures of Centrality:

mean_stars = df['stars'].mean().round(2)
median_stars = df['stars'].median().round(2)
mode_stars = df['stars'].mode().round(2)

display(mean_stars,median_stars, mode_stars)

# Most of the products either have 0/no ratings, so customers usually dont care to rate the products.


#Measures of Dispersion:
variance_stars = df['stars'].var().round(2)
std_dev_stars = df['stars'].std().round(2)
min_stars = df['stars'].min().round(2)
max_stars = df['stars'].max().round(2)
range_stars = (max_price - min_price).round(2)
quantiles_stars = df['stars'].quantile([0.25, 0.5, 0.75])
display(variance_stars, std_dev_stars, min_stars, max_stars, range_stars, quantiles_stars)

# 50% of the customers have given 0 ratings


#Shape of the Distribution:
skewness_stars = df['stars'].skew()
kurtosis_stars = df['stars'].kurtosis()
skewness_stars, kurtosis_stars

#The distribution is almost symmetrical because the skewness is close to 0. the distribution is flatter and less sharply peaked because The kurtosis is negative



#Visualizations:

sns.histplot(df['stars'], kde=True, bins=50, color="salmon")

# the rating "0" is the most common
```
